# <u>Step 4 :- </u> **Incremental Action-Value Update (Alpha / Learning Rate)**

**Why Incremental Update?** :- In Step 2 & 3, we stored all rewards and divided by counts. With incremental updates, we update beliefs on the fly without storing past history.<br>
**Update Formula** :- $Q(a) \leftarrow Q(a) + \alpha \times [R - Q(a)]$ <br>
**Alpha ($\alpha$)** :- Learning rate / step size that controls how much newly observed rewards change our estimate.

## Environment :-


```python
import random

actions = ['left', 'right']

def environment(action):
    if action == 'left':
        return 1 if random.random() < 0.2 else 0
    else:
        return 1 if random.random() < 0.8 else 0
```

## Memory and Parameters :-

Notice: No `reward_count` dictionary is needed anymore! We update `Q` values directly.

```python
Q = {'left': 0, 'right': 0}
action_count = {'left': 0, 'right': 0}

alpha = 0.1
epsilon = 0.1
total_reward = 0
```

## Training :-

### inside the loop :- 

**1. Action Selection :** Explore randomly with probability $\epsilon$, else exploit action with highest $Q$ value <br>
**2. Step Environment :** Call `environment(action)` to receive reward $R$ and update total score <br>
**3. Incremental Update :** $Q(a) \mathrel{+}= \alpha \times (R - Q(a))$

```python
for i in range(10000):

    # Choose an action using epsilon-greedy
    if random.random() < epsilon:
        action = random.choice(actions)  # Explore
    else:
        action = max(Q, key=Q.get)       # Exploit

    reward = environment(action)
    total_reward += reward

    action_count[action] += 1

    # Update the estimated action value
    Q[action] += alpha * (reward - Q[action])
```

## Results after Training :-


```python
print("\nEstimated values:")
for action in actions:
    print(action, round(Q[action], 3))

print("\nAction counts:", action_count)
print("Total reward:", total_reward)
```

# **challenge** : <u>The 3-box reward game (Using Incremental Updates)</u>

Now solving the 3-box problem using our new incremental update rule without `reward_count`!

**BOX A** : <u>10% Chances of Reward </u><br>
**BOX B** : <u>50% Chances of Reward </u><br>
**BOX C** : <u>80% Chances of Reward </u><br>

```python
import random

actions = ['A', 'B', 'C']

def environment(action):
    if action == 'A':
        return 1 if random.random() < 0.1 else 0
    elif action == 'B':
        return 1 if random.random() < 0.5 else 0
    else:
        return 1 if random.random() < 0.8 else 0

Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
action_count = {'A': 0, 'B': 0, 'C': 0}

alpha = 0.1
epsilon = 0.1
total_reward = 0

for i in range(10000):
    if random.random() < epsilon:
        action = random.choice(actions)
    else:
        action = max(Q, key=Q.get)

    reward = environment(action)
    total_reward += reward
    action_count[action] += 1

    # Incremental update
    Q[action] += alpha * (reward - Q[action])

print("Training completed!")
print("Total reward:", total_reward)
print("\nEstimated values:")
for action in actions:
    print(f"{action} Selected: {action_count[action]} times | Estimated value: {round(Q[action], 3)}")
```
