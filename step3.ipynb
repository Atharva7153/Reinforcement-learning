# <u>Step 3 :- </u> **Epsilon Greedy Stratergy (Exploration vs Exploitation)**

**Exploration (ϵ Probability)** :- Try random actions to learn about them <br>
**Exploitation (1 - ϵ Probability)** :- Pick the best action known so far

## Environment :-

```python
import random

actions = ['left', 'right']

def environment(action):
    if action == 'left':
        return 1 if random.random() < 0.2 else 0
    else:
        return 1 if random.random() < 0.8 else 0
```

## Memory and Parameters :-

```python
reward_count = {'left': 0, 'right': 0}
action_count = {'left': 0, 'right': 0}

epsilon = 0.1
total_reward = 0
```

## Training :-

### inside the loop :- 

**1. Calculate Current values :** Q(a) = reward_count[a]/action_count[a] <br>
**2. Action Selection :** if random.random() < epsilon (explore randomly) else pick action with maximum value <br>
**3. Step & update :** Call environment(action), update action_count, reward_count and total_reward <br>

```python
for i in range(10000):

    
    values = {}

    for action in actions:
        if action_count[action] == 0:
            values[action] = 0
        else:
            values[action] = (
                reward_count[action] / action_count[action]
            )


    if random.random() < epsilon:
        action = random.choice(actions)
    else:
        best_value = max(values.values())

        

        
        best_actions = [
            a for a in actions
            if values[a] == best_value
        ]

        action = random.choice(best_actions)

    reward = environment(action)

    action_count[action] += 1
    reward_count[action] += reward
    total_reward += reward
```

## Results after Training :-

```python
print("Training completed!")
print("Total reward:", total_reward)

for action in actions:
    estimated_value = (
        reward_count[action] / action_count[action]
    )

    print(
        action,
        "Selected:", action_count[action],
        "times | Estimated value:",
        round(estimated_value, 3)
    )
```

# **challenge** : <u>The 3-box reward game</u>

Imagine a Ai has 3 boxes. Every time it chooses a box, it gets either 1 reward point or 0 points.

**BOX A** : <u>10% Chances of Reward </u><br>
**BOX B** : <u>50% Chances of Reward </u><br>
**BOX C** : <u>80% Chances of Reward </u><br>

```python
import random

actions = ['A', 'B', 'C']

def environment(action):
    if action == 'A':
        if random.random() < 0.1:
            return 1
        else:
            return 0
    elif action == 'B':
        if random.random() < 0.5:
            return 1
        else:
            return 0
    else:
        if random.random() < 0.8:
            return 1
        else:
            return 0

action_count = {
    'A' : 0,
    'B' : 0,
    'C' : 0
}

reward_count = {
    'A' : 0,
    'B' : 0,
    'C' : 0
}

epsilon = 0.1
total_reward = 0


for i in range(10000):

    values = {}

    for action in actions:
        if action_count[action] == 0:
            values[action] = 0
        else:
            values[action] = reward_count[action]/action_count[action]

    if random.random() < epsilon:
        action = random.choice(actions)
    else:
        best_value = max(values.values())
        best_actions = [
                a for a in actions
                if values[a] == best_value
            ]
        action = random.choice(best_actions)

    reward = environment(action)
    
    action_count[action] += 1
    reward_count[action] += reward
    total_reward += reward

print("Training completed!")
print("Total reward:", total_reward)

for action in actions:
    if action_count[action] == 0:
        estimated_value = 0
    else:
        estimated_value = (
            reward_count[action] / action_count[action]
        )

    print(
        action,
        "Selected:", action_count[action],
        "times | Estimated value:",
        round(estimated_value, 3)
    )
```
